```python

def write_pcd(points, labels, uncertainties, filename):
    num_points = points.shape[0]
    header = [
        "# .PCD v0.7 - Point Cloud Data file format",
        "VERSION 0.7",
        "FIELDS x y z label uncertainty",
        "SIZE 4 4 4 4 4",
        "TYPE F F F U U",
        "COUNT 1 1 1 1 1",
        f"WIDTH {num_points}",
        "HEIGHT 1",
        "VIEWPOINT 0 0 0 1 0 0 0",
        f"POINTS {num_points}",
        "DATA ascii"
    ]
    with open(filename, 'w') as f:
        for line in header:
            f.write(line + "\n")
        for i in range(num_points):
            f.write(f"{points[i,0]} {points[i,1]} {points[i,2]} {labels[i]} {uncertainties[i]}\n")

def convert_ply_to_pcd(mesh_path, output_pcd_path, has_label=True, has_uncertainty=False):
    plydata = PlyData.read(mesh_path)
    vertex = plydata['vertex'].data
    # Extract point coordinates
    points = np.vstack([vertex['x'], vertex['y'], vertex['z']]).T

    # Check and extract the label field
    if 'label' in vertex.dtype.names:
        labels = vertex['label']
    else:
        raise ValueError("The PLY file does not contain a 'label' property.")
    
    if 'uncertainty' in vertex.dtype.names:
        uncertainties = vertex['uncertainty']
    else:
        uncertainties = np.zeros_like(labels)

    write_pcd(points, labels, uncertainties, output_pcd_path)
    
```
